Full chart pack is in `reports/stage4_eda_report.pdf`. Every chart has a written observation in `reports/stage4_eda_summary.md`.



```python
import sys
from pathlib import Path
ROOT = Path.cwd() if (Path.cwd() / 'run_phase1.py').exists() else Path.cwd().parent
sys.path.insert(0, str(ROOT))
import pandas as pd
import matplotlib.pyplot as plt
from src.utils.config import PROCESSED_DIR, REPORTS_DIR

df = pd.read_csv(PROCESSED_DIR / 'processed_sensor_data.csv', parse_dates=['timestamp'])
daily = df.groupby(df['timestamp'].dt.date)['energy_kwh'].sum()
ax = daily.plot(figsize=(10,4), title='Daily campus energy (kWh)')
plt.tight_layout()
print((REPORTS_DIR / 'stage4_eda_summary.md').read_text(encoding='utf-8')[:1500])

```
